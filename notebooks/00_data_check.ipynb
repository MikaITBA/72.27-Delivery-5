# Data check

Load `font.h`, plot all 32 letters and a noisy version (p = 0.1).

```python
import numpy as np

from tp5.data import add_noise, load_font, pixel_errors, show_letters
```

```python
X, chars = load_font()
print(X.shape)
print(chars)
```

```python
show_letters(X, titles=chars);
```

## Noise example (p = 0.1)

```python
rng = np.random.default_rng(0)
X_noisy = add_noise(X, 0.1, rng=rng)
errors = pixel_errors(X, X_noisy)
print(f"flipped pixels per letter: mean {errors.mean():.2f}, max {errors.max()}")
show_letters(X_noisy, titles=[f"{c} ({e})" for c, e in zip(chars, errors)]);
```
